```python
# Program to handle ZeroDivisionError when dividing a number by zero

try:
    # Taking input from user
    num = int(input("Enter a number: "))
    
    # Attempting to divide the number by zero (which will raise an exception)
    result = num / 0  

# Catching the ZeroDivisionError exception
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
```
