```python
# Program to handle TypeError when non-numerical input is entered

try:
    # Taking two numerical inputs from the user
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    
    # Performing addition and displaying the result
    print(f"Sum: {num1 + num2}")

# Catching ValueError if the inputs are not numerical
except ValueError:
    print("Error: Please enter valid numerical values.")
```
