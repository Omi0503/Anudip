```python
# Program to handle FileNotFoundError when trying to open a non-existent file

try:
    # Taking file name input from the user
    filename = input("Enter the file name: ")
    
    # Attempting to open the file in read mode
    with open(filename, "r") as file:
        # Reading the file content and displaying it
        content = file.read()
        print(content)

# Catching FileNotFoundError if the file does not exist
except FileNotFoundError:
    print("Error: The file does not exist.")
```
