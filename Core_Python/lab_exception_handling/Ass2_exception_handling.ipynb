```python
# Program to handle ValueError if input is not a valid integer

try:
    # Taking input from the user and trying to convert it to an integer
    num = int(input("Enter an integer: "))
    
    # Printing the valid integer input
    print(f"You entered: {num}")

# Catching ValueError if the input is not a valid integer
except ValueError:
    print("Error: Invalid input. Please enter a valid integer.")
```
