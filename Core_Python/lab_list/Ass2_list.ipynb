```python
# Write a Python program to get the largest and smallest number from a list without builtin functions.

my_list = [4, 2, 5, 1, 6, 9, 7, 2]
min_num = my_list[0]
max_num = my_list[0]

for num in my_list:
    if num < min_num:
        min_num = num
    elif num > max_num:
        max_num = num
print(f'min num from the list = {min_num}\nmax num from the list = {max_num}')
```
