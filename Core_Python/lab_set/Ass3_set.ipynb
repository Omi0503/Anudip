```python
# Define two sets with some common and different elements
set1 = {10, 20, 30, 40, 50}
set2 = {60, 70, 80, 90, 10}

# Find the common elements between set1 and set2
common_elements = set1.intersection(set2)

# Check if there are any common elements
if common_elements:
    # If common elements are found, print them
    print("Common Elements:", common_elements)
else:
    # If no common elements are found, print a message
    print("No common elements found.")
```
