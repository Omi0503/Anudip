```python
# Define the first set with some numbers
set1 = {10, 20, 30, 40, 50}  

# Define the second set with some overlapping and unique numbers
set2 = {30, 40, 50, 60, 70}  

# Perform union operation to get all unique elements from both sets
unique_items = set1.union(set2)  

# Print the result containing all unique elements
print("Unique Items:", unique_items)  
```
