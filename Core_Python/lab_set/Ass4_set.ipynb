```python
# Define two sets with some common and different elements
set1 = {10, 20, 30, 40, 50}
set2 = {30, 40, 50, 60, 70}

# Update set1 to keep only the elements that are common between set1 and set2
set1.intersection_update(set2)

# Print the updated set1, which now contains only the common elements
print("Updated Set1:", set1)
```
