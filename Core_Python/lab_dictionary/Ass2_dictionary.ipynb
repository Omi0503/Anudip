```python
# Define three dictionaries with key-value pairs
dict1 = {1: 10, 2: 20}
dict2 = {3: 30, 4: 40}
dict3 = {5: 50, 6: 60}

# Concatenate the dictionaries using unpacking
combined_dict = {**dict1, **dict2, **dict3}

# Print the combined dictionary containing all the key-value pairs from the three dictionaries
print("Combined Dictionary:", combined_dict)
```
