```python
# Define a dictionary with some empty (None) values
input_dict = {1: 10, 2: 20, 3: None, 4: 40, 5: None, 6: 60}

# Create a new dictionary by filtering out the items with None values
filtered_dict = {key: value for key, value in input_dict.items() if value is not None}

# Print the filtered dictionary with empty (None) items removed
print("Dictionary with Empty Items Dropped:", filtered_dict)
```
