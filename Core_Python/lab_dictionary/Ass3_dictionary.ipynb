```python
# Define a dictionary with key-value pairs
dict_num = {1: 10, 2: 20, 3: 30, 4: 40, 5: 50, 6: 60}

# Print header for the table of keys and values
print("key\tvalue")

# Iterate through the dictionary using the items() method to get both key and value
for key, value in dict_num.items():
    # Print each key-value pair, formatted with a tab space between them
    print(f"{key}\t{value}")
```
