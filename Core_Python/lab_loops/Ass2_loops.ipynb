```python
#Write a python program to check whether a number is palindrome or not?
```

```python
num=int(input("Enter number:"))
temp=num
rev=0
while(num>0):
    dig=num%10
    rev=rev*10+dig
    num=num//10
if(temp==rev):
    print("The number is a palindrome!")
else:
    print("The number not a palindrome!")
```
