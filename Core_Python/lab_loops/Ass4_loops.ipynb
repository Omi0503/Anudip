```python
#Accept numbers using input() function until the user enters 0. If user input 0 then break the while loop and display the sum of all the numbers
```

```python
total_sum = 0
while True:
    num = int(input("Enter a number: "))
    if num == 0:
        break 
    total_sum += num
print("The sum of all numbers is:", total_sum)
```

```python
# usingWalrus Operator (:=)
```

```python
total_sum = 0
while  (num := int(input("Enter a number: "))) != 0:
    total_sum += num
print("The sum of all numbers is:", total_sum)
```
