```python
# Importing numpy library
import numpy as np
# Defining the list of numbers
my_list = [1, 2, 3, 4, 5]
# Converting the list into a numpy array
numpy_array = np.array(my_list)
# Displaying the first and last element of the array
print("First element:", numpy_array[0])
print("Last element:", numpy_array[-1])
# Doubling each element in the array and displaying the result
doubled_array = numpy_array * 2
print("Array after doubling each element:")
print(doubled_array)
```
