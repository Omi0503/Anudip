```python
# Q. Splitting customer data based on last purchase days

import numpy as np

# Given customer IDs and last purchase days
customer_ids = np.array([101, 102, 103, 104, 105, 106, 107, 108, 109, 110])
last_purchase_days_ago = np.array([5, 15, 20, 25, 30, 35, 40, 45, 50, 55])

# Identifying active customers (purchase in the last 30 days)
active_customers = customer_ids[last_purchase_days_ago <= 30]

# Identifying inactive customers (no purchase in the last 30 days)
inactive_customers = customer_ids[last_purchase_days_ago > 30]

print("Active Customers (Last purchase <= 30 days ago):\n", active_customers,"\n")
print("Inactive Customers (Last purchase > 30 days ago):\n", inactive_customers)
```
