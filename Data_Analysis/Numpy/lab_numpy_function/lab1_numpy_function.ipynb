```python
import numpy as np

# Given temperature data
temperatures = np.array([32.5, 34.2, 36.8, 29.3, 31.0, 38.7, 23.1, 16.8, 22.8, 37.2, 4.0, -4.0, -12.0])

# Identify hot days (temperature > 35 degrees Celsius)
print("Hot Days:")
print("Day     Temperature")
for i in range(len(temperatures)):
    if temperatures[i] > 35:
        print(f"{i+1:<8}{temperatures[i]}")

# Identify cold days (temperature < 5 degrees Celsius)
print("\nCold Days:")
print("Day     Temperature")
for i in range(len(temperatures)):
    if temperatures[i] < 5:
        print(f"{i+1:<8}{temperatures[i]}")
```
