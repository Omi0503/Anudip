```python
# Q. Combine employee data

import numpy as np

# Full-time employees dataset
full_time_employees = np.array([
    [101, "John Doe", "Full-Time", 55000],
    [102, "Jane Smith", "Full-Time", 60000],
    [103, "Mike Johnson", "Full-Time", 52000]
])

# Part-time employees dataset
part_time_employees = np.array([
    [201, "Alice Brown", "Part-Time", 25000],
    [202, "Bob Wilson", "Part-Time", 28000],
    [203, "Emily Davis", "Part-Time", 22000]
])

# Combine full-time and part-time employees
total_employees = np.vstack((full_time_employees, part_time_employees))

print("\nCombined Employee Data:")
for emp in total_employees:
    print(f" Employee ID: {int(emp[0])} , Name: {emp[1]} , Type: {emp[2]} , Salary: {int(emp[3])}")
```
