```python
# A Subplot allows you to create multiple plots within the same figure. 
# This is useful when you want to compare different datasets or visualizations side by side, without creating separate figures for each plot.

# The subplot function in Matplotlib divides the figure into a grid of subplots and places each plot in the specified grid location.

# Syntax of subplot:
                    
#                 plt.subplot(nrows, ncols, index)

    
#     nrows: Number of rows in the grid.
#     ncols: Number of columns in the grid.
#     index: The index of the subplot where the plot will appear. Index starts from 1 (top-left corner).
```

```python
# Example1:
# Q. input are name of student and marks of two subject 

# Creates a 2x1 grid of subplots (2 rows, 1 column).
# fig is the figure object, and axs is an array of axes (subplots).
# figsize=(12,5) sets the size of the overall figure.

# Expected Output:
#     First Subplot (axs[0]): A blue bar chart showing the math scores of students.
#     Second Subplot (axs[1]): A green bar chart showing the science scores of students.
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Sample student data
students = ['Jhon', 'Smith', 'Marry', 'Rose', 'Devid','Jhon1', 'Smith1', 'Marry1', 'Rose1', 'Devid1']
math_scores = [85, 92, 78, 88, 90,85, 92, 78, 88, 90]
science_scores = [76, 88, 92, 80, 78,76, 88, 92, 80, 78]

# Create a subplot with two subplots (2 row, 1 columns)
fig, axs = plt.subplots(2,1, figsize=(12,5))

# Plot Math Scores
axs[0].bar(students, math_scores, color='b')
axs[0].set_title('Math Scores')
axs[0].set_xlabel('Students')
axs[0].set_ylabel('Score')

# Plot Science Scores
axs[1].bar(students, science_scores, color='g')
axs[1].set_title('Science Scores')
axs[1].set_xlabel('Students')
axs[1].set_ylabel('Score')

# Adjust spacing between subplots
plt.tight_layout()    #used to automatically adjust the spacing between the subplots without overlapping.

# Show the subplot
plt.show()
```

```python
# Example3:

# Expected Layout:
#     Top-left: Math Scores (First Plot)
#     Top-right: Science Scores (First Plot)
#     Bottom-left: Math Scores Again (Second Plot)
#     Bottom-right: Science Scores Again (Second Plot)

# This approach will create a 2x2 grid with 4 subplots, each displaying the respective data for math and science scores.
```

```python
import matplotlib.pyplot as plt

# Sample student data
students = ['Jhon', 'Smith', 'Marry', 'Rose', 'Devid','Jhon1', 'Smith1', 'Marry1', 'Rose1', 'Devid1']
math_scores = [85, 92, 78, 88, 90, 85, 92, 78, 88, 90]
science_scores = [76, 88, 92, 80, 78, 76, 88, 92, 80, 78]

# Create a 2x2 subplot grid
fig, axs = plt.subplots(2, 2, figsize=(12, 8))

# Plot Math Scores in the first subplot (top-left)
axs[0, 0].bar(students, math_scores, color='b')
axs[0, 0].set_title('Math Scores')
axs[0, 0].set_xlabel('Students')
axs[0, 0].set_ylabel('Score')

# Plot Science Scores in the second subplot (top-right)
axs[0, 1].bar(students, science_scores, color='g')
axs[0, 1].set_title('Science Scores')
axs[0, 1].set_xlabel('Students')
axs[0, 1].set_ylabel('Score')

# Plot Math Scores (again) in the third subplot (bottom-left)
axs[1, 0].bar(students, math_scores, color='b')
axs[1, 0].set_title('Math Scores Again')
axs[1, 0].set_xlabel('Students')
axs[1, 0].set_ylabel('Score')

# Plot Science Scores (again) in the fourth subplot (bottom-right)
axs[1, 1].bar(students, science_scores, color='g')
axs[1, 1].set_title('Science Scores Again')
axs[1, 1].set_xlabel('Students')
axs[1, 1].set_ylabel('Score')

# Automatically adjust the layout to prevent overlap
plt.tight_layout()

# Show the plot
plt.show()
```

```python
# Example4:
# Q. How do you create a subplot with three bar charts in a single row, displaying the math, science, and Java scores of students?

# Output:
#     This will display a figure with three side-by-side bar charts:
#         First chart: Math Scores (in blue).
#         Second chart: Science Scores (in green).

# Summary of Key Concepts:
#     plt.subplots(1, 3): Creates a subplot with 1 row and 3 columns.
#     axs: Array containing axes for each subplot.
#     tight_layout(): Adjusts layout to ensure subplots do not overlap.
#     Bar charts are used to visually represent student scores in Math, Science, and Java.
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Sample student data
students = ['Jhon', 'Smith', 'Marry', 'Rose', 'Devid']  # List of student names
math_scores = [85, 92, 78, 88, 90]  # Math scores for each student
science_scores = [76, 88, 92, 80, 78]  # Science scores for each student
Java_scores = [45, 88, 20, 57, 90]  # Java scores for each student

# Create a subplot with three subplots (1 row, 3 columns) and set figure size
fig, axs = plt.subplots(1, 3, figsize=(15, 5))

# Plot Math Scores
axs[0].bar(students, math_scores, color='b')  # Create a bar chart for Math scores in blue
axs[0].set_title('Math Scores')  # Set the title for the first subplot
axs[0].set_xlabel('Students')  # Set the label for the x-axis
axs[0].set_ylabel('Score')  # Set the label for the y-axis

# Plot Science Scores
axs[1].bar(students, science_scores, color='g')  # Create a bar chart for Science scores in green
axs[1].set_title('Science Scores')  
axs[1].set_xlabel('Students')  
axs[1].set_ylabel('Score') 

# Plot Java Scores
axs[2].bar(students, Java_scores, color='r')  # Create a bar chart for Java scores in red
axs[2].set_title('Java Scores')
axs[2].set_xlabel('Students')
axs[2].set_ylabel('Score')

# Adjust spacing between subplots to ensure no overlap
plt.tight_layout()

# Show the final plot with all subplots
plt.show()
```

```python
# Example5:
# Q. This example demonstrates how to create a simple 2x2 grid of subplots.

# Explanation:
#     plt.subplot(2, 2, 1) creates the first subplot in a 2x2 grid (top-left).
#     plt.subplot(2, 2, 2) creates the second subplot (top-right), and so on.
#     plt.tight_layout() automatically adjusts the subplot parameters to avoid overlapping text or labels.
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data for plotting
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create a 2x2 grid of subplots
plt.subplot(2, 2, 1)  # 2 rows, 2 columns, first subplot
plt.plot(x, y)
plt.title('Sine Wave')

plt.subplot(2, 2, 2)  # 2 rows, 2 columns, second subplot
plt.plot(x, np.cos(x))
plt.title('Cosine Wave')

plt.subplot(2, 2, 3)  # 2 rows, 2 columns, third subplot
plt.plot(x, np.tan(x))
plt.title('Tangent Wave')
plt.ylim(-10, 10)  # Limit the y-axis for better visualization of tangent

plt.subplot(2, 2, 4)  # 2 rows, 2 columns, fourth subplot
plt.plot(x, np.exp(x / 10))
plt.title('Exponential')

# Adjust the layout to prevent overlap
plt.tight_layout()

# Show the plot
plt.show()
```

```python
# Example6:
# Q. How do you create a 2x2 grid of subplots in Matplotlib,plot different variations of a sine function,set axis labels and subplot's title?

# Output:
#     You will see a 2x2 grid of subplots, where:
#         The top-left subplot displays the plot for y = sin(x^2).
#         The top-right subplot displays the same function in red.
#         The bottom-left subplot displays the negative of the function y = -sin(x^2) in green.
#         The bottom-right subplot displays the sine function with negative x-values y = sin(-x) in orange.
#     Each subplot will have its own title and will share the common x-axis and y-axis labels ("xLabel" and "yLabel").
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate data
x = np.linspace(0, 2*np.pi, 400)  # Create an array of 400 points from 0 to 2π
y = np.sin(x**2)  # Compute sine of x squared for each point in x

# Create a 2x2 grid of subplots
fig, axs = plt.subplots(2, 2)

# Plot on the first subplot (1st row, 1st column)
axs[0, 0].plot(x, y)  # Plot y = sin(x^2) on the first axis
axs[0, 0].set_title("Axis 1st row 1st col")  # Set title for the first subplot

# Plot on the second subplot (1st row, 2nd column) with a red line
axs[0, 1].plot(x, y, 'tab:red')  # Use red color for the plot
axs[0, 1].set_title("Axis 1st row 2nd col")

# Plot on the third subplot (2nd row, 1st column) with negative sine values (green)
axs[1, 0].plot(x, -y, 'tab:green')  # Negative sine values (in green)
axs[1, 0].set_title("Axis 2nd row 1st col") 

# Plot on the fourth subplot (2nd row, 2nd column) with negative x values (orange)
axs[1, 1].plot(-x, y, 'tab:orange')  # Negative x-values (in orange)
axs[1, 1].set_title("Axis 2nd row 2nd col")

# Set labels for all subplots
for ax in axs.flat:  # Loop through all subplots
    ax.set(xlabel='xLabel', ylabel='yLabel')

# Hide x and y ticks on the outer subplots to prevent overlap
for ax in axs.flat:  
    ax.label_outer()  # Remove x and y ticks for outer subplots to make it cleaner

# Adjust layout to prevent overlap and display the plot
plt.tight_layout()  

plt.show()  # Show the final plot
```

```python
# Example7: Sharing Axes in Subplots

# Explanation:
#     fig, axes = plt.subplots(2, 1, sharex=True) creates a 2x1 grid of subplots where the x-axis is shared between the two subplots. 
#     This makes it easier to compare both graphs, as the x-axis is aligned.
```

```python
# Create a 2x1 grid of subplots sharing the x-axis
fig, axes = plt.subplots(2, 1, sharex=True)

# Plot the first subplot
axes[0].plot(x, np.sin(x), color='b')
axes[0].set_title('Sine Wave')

# Plot the second subplot
axes[1].plot(x, np.cos(x), color='r')
axes[1].set_title('Cosine Wave')

# Add labels for the x-axis and y-axis
for ax in axes:
    ax.set_ylabel('Value')

# Show the plot
plt.tight_layout()
plt.show()
```

```python
# Example8: Customizing Subplots Layout

# Explanation:
#     fig, axes = plt.subplots(1, 3, figsize=(12, 4)) creates a 1-row and 3-column subplot with a custom figure size.
#     plt.tight_layout(pad=2.0) adjusts the padding between subplots to avoid overlap.
#     You can customize the figure size using the figsize parameter.
```

```python
# Create a 1x3 grid of subplots with a specific size
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

# Plot each subplot
axes[0].plot(x, np.sin(x), 'g')
axes[0].set_title('Sine Wave')

axes[1].plot(x, np.cos(x), 'r')
axes[1].set_title('Cosine Wave')

axes[2].plot(x, np.tan(x), 'b')
axes[2].set_title('Tangent Wave')
axes[2].set_ylim(-10, 10)  # Limit the y-axis for better visualization

# Adjust the layout to ensure there's space between subplots
plt.tight_layout(pad=2.0)

# Show the plot
plt.show()
```

```python
# Example9: Using add_subplot() Method

# Alternatively, you can create subplots using the add_subplot() method of a figure object.

# Explanation:
#     fig.add_subplot(2, 2, 1) adds a subplot in a 2x2 grid. This is an alternative to plt.subplot() but allows more control over the figure object.
#     fig.tight_layout() ensures that the subplots do not overlap.
```

```python
# Create a figure
fig = plt.figure(figsize=(10, 6))

# Add the first subplot
ax1 = fig.add_subplot(2, 2, 1)
ax1.plot(x, np.sin(x))
ax1.set_title('Sine Wave')

# Add the second subplot
ax2 = fig.add_subplot(2, 2, 2)
ax2.plot(x, np.cos(x))
ax2.set_title('Cosine Wave')

# Add the third subplot
ax3 = fig.add_subplot(2, 2, 3)
ax3.plot(x, np.tan(x))
ax3.set_title('Tangent Wave')
ax3.set_ylim(-10, 10)  # Limit the y-axis for better visualization

# Add the fourth subplot
ax4 = fig.add_subplot(2, 2, 4)
ax4.plot(x, np.exp(x / 10))
ax4.set_title('Exponential')

# Adjust layout
fig.tight_layout()

# Show the plot
plt.show()
```

```python
# Example10: Subplots with Different Row and Column Sizes
# You can also customize the row and column sizes by using gridspec to control the subplot sizes more precisely.

# Explanation:
#     gridspec.GridSpec(2, 2, width_ratios=[3, 1], height_ratios=[1, 2]) defines a 2x2 grid with customized row and column sizes.
#     The first column is 3 times as wide as the second column.
#     The second row is 2 times as tall as the first row.
#     This allows you to have more flexible control over the subplot sizes.
```

```python
import matplotlib.gridspec as gridspec

# Create a figure
fig = plt.figure(figsize=(10, 6))

# Define a GridSpec
gs = gridspec.GridSpec(2, 2, width_ratios=[3, 1], height_ratios=[1, 2])

# Add subplots to specific grid locations
ax1 = fig.add_subplot(gs[0])  # Top-left subplot (larger width)
ax1.plot(x, np.sin(x))
ax1.set_title('Sine Wave')

ax2 = fig.add_subplot(gs[1])  # Top-right subplot (smaller width)
ax2.plot(x, np.cos(x))
ax2.set_title('Cosine Wave')

ax3 = fig.add_subplot(gs[2])  # Bottom-left subplot (larger height)
ax3.plot(x, np.tan(x))
ax3.set_title('Tangent Wave')
ax3.set_ylim(-10, 10)  # Limit the y-axis for better visualization

# Adjust layout
fig.tight_layout()

# Show the plot
plt.show()
```
