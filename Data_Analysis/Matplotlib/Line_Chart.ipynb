```python
#Example Score Chart
#Marker = 'D', linestyle='-.', color='c', label="Stock Price"


import matplotlib.pyplot as plt
x=[101,102,103]
y=[350,500,467]
plt.plot(x,y,marker='D',linestyle='-.',color='black',ms=4,mec='red',mfc='white')
plt.xlabel("Reg No.")
plt.ylabel("Scores")
plt.title("Score Chart")
plt.legend(["Line"])
plt.show()
```

```python
#Example Stock Price Trend Over 10 Days
#Marker = 's', linestyle='-', color='c', label="Stock Price"

import matplotlib.pyplot as plt
import numpy as np

# Sample stock price data
days = np.arange(1, 11)
prices = [100, 102, 98, 105, 110, 108, 115, 118, 120, 125]

plt.plot(days, prices,marker='s', linestyle='-', color='c', label="Stock Price")
plt.xlabel("Days")
plt.ylabel("Price (USD)")
plt.title("Stock Price Trend Over 10 Days")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Example Sales of two store A and B

import matplotlib.pyplot as plt

# Sales data for two stores (in thousands)
categories = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
store_a_sales = [15, 20, 30, 25, 18, 22]
store_b_sales = [12, 18, 25, 20, 22, 24]

# Plotting the sales data for Store A and Store B
plt.plot(categories, store_a_sales, marker='o', label='Store A', color='blue')
plt.plot(categories, store_b_sales, marker='s', label='Store B', color='green')

# Adding labels and title
plt.xlabel('Month')
plt.ylabel('Sales (in thousands)')
plt.title('Sales Comparison of Store A and Store B')

# Adding a legend
plt.legend()

# Display the plot
plt.show()
```
