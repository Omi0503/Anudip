```python
# Pie Chart (plt.pie()) :

# Used to show percentage distribution of categories.

#  Use case: Market share, budget allocation, election results.
```

```python
import matplotlib.pyplot as plt

categories = ['Apple', 'Banana', 'Cherry', 'Date']
values = [30, 20, 40, 10]

# Create Pie Chart
plt.pie(values, labels=categories, autopct='%1.1f%%', startangle=90, colors=['red', 'yellow', 'pink', 'brown'])

plt.title("Fruit Distribution")
plt.show()
```

```python
# Key plt.pie() Parameters:

# Parameter	:  Description
# labels	    :  Labels for each slice
# sizes	    :  Values that determine slice size
# colors	    :  Set custom colors
# autopct	    :  Show percentage labels ('%1.1f%%')
# explode	    :  Highlight a slice ((0, 0.1, 0, 0))
# shadow	    :  Add shadow effect (True/False)
# startangle	:  Rotate the pie chart
# wedgeprops	:  Customize slice edges (e.g., {'edgecolor': 'black'})
```

```python
# Basic Example of plt.pie()
# plt.pie() is a function in Matplotlib used to create pie charts.

# Explanation:
# sizes: Values corresponding to each slice.
# labels: Names for the slices.
# autopct='%1.1f%%': Displays percentage values on slices.
# startangle=90: Rotates the chart for better visualization.
```

```python
import matplotlib.pyplot as plt

# Data
labels = ['Apple', 'Banana', 'Cherry', 'Date']
sizes = [30, 20, 40, 10]  # Percentage distribution

# Create Pie Chart
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90)

# Title
plt.title("Fruit Distribution")

# Show the plot
plt.show()
```

```python
# 1. Adding Colors and Styling

# Additional Features:

# colors: Manually set slice colors.
# explode: Pulls a slice out for emphasis.
# shadow=True: Adds a shadow effect.
```

```python
import matplotlib.pyplot as plt

colors = ['red', 'yellow', 'pink', 'brown']
explode = (0, 0.1, 0, 0)  # "explode" Banana slice

plt.pie(sizes, labels=labels, autopct='%1.1f%%', colors=colors, 
        explode=explode, shadow=True, startangle=140)

plt.title("Fruit Distribution with Colors & Shadow")
plt.show()
```

```python
# 2. Creating a Donut Chart

#  Donut Effect:

# Adds a white circle in the center using plt.Circle().
```

```python
import matplotlib.pyplot as plt

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, wedgeprops={'edgecolor': 'black'})

# Draw a circle in the middle to create a donut effect
plt.gca().add_artist(plt.Circle((0, 0), 0.5, color='white'))

plt.title("Donut Chart")
plt.show()
```

```python
# 3. Pie Chart with Percentage Formatting

# Custom Label Formatting:

# Shows percentage and actual values inside slices.
```

```python
import matplotlib.pyplot as plt

def custom_autopct(pct):
    return f'{pct:.2f}% ({int(pct * sum(sizes) / 100)})'  # Shows percentage & actual count

plt.pie(sizes, labels=labels, autopct=custom_autopct, startangle=90)
plt.title("Pie Chart with Custom Percentage Display")
plt.show()
```

```python
# 📌 Basic Example Using plt.pie()
```

```python
import matplotlib.pyplot as plt

# Data
labels = ['Apple', 'Banana', 'Cherry', 'Date']
sizes = [30, 20, 40, 10]  # Percentage distribution
colors = ['red', 'yellow', 'pink', 'brown']
explode = (0, 0.1, 0, 0)  # "explode" Banana slice

# Create Pie Chart
plt.pie(sizes, labels=labels, autopct='%1.1f%%', colors=colors, 
        explode=explode, shadow=True, startangle=140, 
        wedgeprops={'edgecolor': 'black'})

# Title
plt.title("Fruit Distribution")

# Show the plot
plt.show()
```

```python
# Explanation of Parameters with Examples :

# Parameter	    : Description	                     : Example Usage

# labels	    : Labels for each slice	             : labels=['Apple', 'Banana', 'Cherry']
# sizes	        : Values that determine slice size	 : sizes=[30, 20, 40]
# colors	    : Set custom colors	                 : colors=['red', 'yellow', 'pink']
# autopct	    : Show percentage labels on slices	 : autopct='%1.1f%%'
# explode	    : Pull a slice out for emphasis	     : explode=(0, 0.1, 0, 0) (Banana slice pulled out)
# shadow	    : Add shadow effect	                 : shadow=True
# startangle	: Rotate the pie chart	             : startangle=140 (Rotates 140°)
# wedgeprops	: Customize slice edges	             : wedgeprops={'edgecolor': 'black'} 
```

```python
#  1. Using autopct to Show Percentage Values
#  This displays percentage values inside each slice.
```

```python
import matplotlib.pyplot as plt

plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title("Pie Chart with Percentage Labels")
plt.show()
```

```python
import matplotlib.pyplot as plt

explode = (0, 0.1, 0, 0)  # Pull the second slice out
plt.pie(sizes, labels=labels, explode=explode)
plt.title("Pie Chart with Exploded Slice")
plt.show()
```

```python
# 4. Using shadow to Add a 3D Effect
# This adds a shadow effect for better visualization.
```

```python
#.📌Example 1: Basic Pie Chart with Shadow

# 🔹Effect: A simple pie chart with a shadow underneath.
```

```python
import matplotlib.pyplot as plt

# Data
labels = ['Apple', 'Banana', 'Cherry', 'Date']
sizes = [30, 20, 40, 10]

# Create Pie Chart with Shadow
plt.pie(sizes, labels=labels, shadow=True)

# Title
plt.title("Basic Pie Chart with Shadow")

# Show plot
plt.show()
```

```python
# 📌 Example 2: Shadow with Exploded Slice
# 🔹 Effect: The Banana slice is pulled out and has a shadow.
```

```python
# Data
colors = ['red', 'yellow', 'pink', 'brown']
explode = (0, 0.1, 0, 0)  # Highlight Banana slice

# Pie Chart with Explode and Shadow
plt.pie(sizes, labels=labels, autopct='%1.1f%%', colors=colors,
        explode=explode, shadow=True, startangle=90)

plt.title("Exploded Pie Chart with Shadow")
plt.show()
```

```python
# 📌 Example 3: Shadow with Custom Wedge Borders
# 🔹 Effect: Each slice has a black border, making it more visually distinct.
```

```python
# Pie Chart with Custom Edge Properties
plt.pie(sizes, labels=labels, autopct='%1.1f%%', shadow=True, startangle=140,
        wedgeprops={'edgecolor': 'black', 'linewidth': 2})

plt.title("Pie Chart with Shadow and Black Borders")
plt.show()
```

```python
# 📌 Example 4: Multi-Color Pie Chart with Shadow
# 🔹 Effect: A colorful pie chart with randomly generated colors and a shadow.
```

```python
import numpy as np

# Generate random data
np.random.seed(42)
sizes = np.random.randint(10, 50, size=6)
labels = ['A', 'B', 'C', 'D', 'E', 'F']
colors = plt.cm.Paired(np.linspace(0, 1, len(labels)))  # Generate distinct colors

# Pie Chart
plt.pie(sizes, labels=labels, autopct='%1.1f%%', colors=colors, shadow=True)

plt.title("Multi-Color Pie Chart with Shadow")
plt.show()
```

```python
# 📌 Example 5: Donut Chart with Shadow
# 🔹 Effect: A donut chart with a shadow.
```

```python
# Create Pie Chart
plt.pie(sizes, labels=labels, autopct='%1.1f%%', shadow=True, startangle=90)

# Draw a white circle in the center to create a donut effect
plt.gca().add_artist(plt.Circle((0, 0), 0.5, color='white'))

plt.title("Donut Chart with Shadow")
plt.show()
```

```python
# 5. Using startangle to Rotate the Chart
# This rotates the chart 90°.
```

```python
import matplotlib.pyplot as plt

plt.pie(sizes, labels=labels, startangle=90)
plt.title("Pie Chart Rotated by 90°")
plt.show()
```

```python
# 6. Using wedgeprops to Customize Slice Edges
# This sets black slice edges with thicker borders.
```

```python
import matplotlib.pyplot as plt

plt.pie(sizes, labels=labels, wedgeprops={'edgecolor': 'black', 'linewidth': 2})
plt.title("Pie Chart with Custom Edge Properties")
plt.show()
```
