```python
# 1. Basic Histogram
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate random data
data = np.random.randn(1000)  # 1000 random values from a normal distribution

# Create histogram
plt.hist(data, bins=30, color='blue', edgecolor='black', alpha=0.7)

# Labels and title
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Basic Histogram")

# Show the plot
plt.show()
```

```python
# 2. Histogram with Different Bin Sizes
# Uses two histograms on the same plot to compare different bin sizes.
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.randn(1000)

plt.hist(data, bins=10, color='red', alpha=0.6, label="10 Bins")
plt.hist(data, bins=30, color='green', alpha=0.6, label="30 Bins")

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram with Different Bin Sizes")
plt.legend()
plt.show()
```

```python
# 3. Normalized Histogram (Density Plot)
# density=True makes the histogram represent a probability distribution rather than raw counts.
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.randn(1000)

plt.hist(data, bins=25, density=True, color='purple', alpha=0.6, edgecolor='black')

plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Normalized Histogram (Density Plot)")

plt.show()
```

```python
# 4. Multiple Data Sets in One Histogram
# Two overlapping histograms with transparency for comparison.
```

```python
import numpy as np
import matplotlib.pyplot as plt

data1 = np.random.randn(1000)
data2 = np.random.randn(1000) + 2  # Shifted dataset

plt.hist(data1, bins=30, alpha=0.5, label="Dataset 1", color='blue', edgecolor='black')
plt.hist(data2, bins=30, alpha=0.5, label="Dataset 2", color='orange', edgecolor='black')

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Overlapping Histograms")
plt.legend()
plt.show()
```

```python
# 5. Stacked Histogram
# Uses stacked=True to visualize two distributions cumulatively.
```

```python
import numpy as np
import matplotlib.pyplot as plt

data1 = np.random.randn(1000)
data2 = np.random.randn(1000) + 2  # Shifted dataset

plt.hist([data1, data2], bins=30, stacked=True, color=['blue', 'orange'], label=['Dataset 1', 'Dataset 2'])

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Stacked Histogram")
plt.legend()
plt.show()
```

```python
# 11. Side-by-Side Histograms
# Uses histtype='bar' to plot two histograms side by side instead of overlapping.
```

```python
# 13. Histogram with Error Bars
# Adds error bars to the histogram, useful for showing uncertainty.
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.randn(1000)
counts, bins, patches = plt.hist(data, bins=30, color='green', edgecolor='black', alpha=0.7)

# Calculate error bars (square root of count)
error = np.sqrt(counts)

# Plot error bars
bin_centers = 0.5 * (bins[:-1] + bins[1:])
plt.errorbar(bin_centers, counts, yerr=error, fmt='o', color='red')

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram with Error Bars")

plt.show()
```

```python
# 14. Colored Histogram Based on Frequency
# Changes bar colors dynamically based on frequency.
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.randn(1000)
counts, bins, patches = plt.hist(data, bins=30, edgecolor='black')

# Color bars based on frequency
for patch, count in zip(patches, counts):
    if count < 10:
        patch.set_facecolor('red')
    elif count < 30:
        patch.set_facecolor('yellow')
    else:
        patch.set_facecolor('green')

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Colored Histogram Based on Frequency")

plt.show()
```

```python
# 15. 2D Histogram (Heatmap)
# plt.hist2d() creates a 2D histogram (heatmap).
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.random.randn(1000)
y = np.random.randn(1000)

plt.hist2d(x, y, bins=30, cmap='plasma')

plt.colorbar(label="Frequency")
plt.xlabel("X Value")
plt.ylabel("Y Value")
plt.title("2D Histogram (Heatmap)")

plt.show()
```

```python
# 17. Polar Histogram
# Uses polar coordinates to create a circular histogram.
```

```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.random.uniform(0, 2*np.pi, 1000)

plt.figure(figsize=(6, 6))
ax = plt.subplot(111, projection='polar')
ax.hist(theta, bins=30, color='teal', edgecolor='black')

ax.set_title("Polar Histogram")
plt.show()
```
