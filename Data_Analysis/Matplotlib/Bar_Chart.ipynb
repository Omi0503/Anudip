```python
#Example Grouped Bar Chart with 20 Bars per Category

import matplotlib.pyplot as plt
import numpy as np

# Data
categories = ['A', 'B', 'C']
n_bars = 20
group_values = np.random.randint(1, 20, (n_bars, len(categories)))  # Generate random values for 20 groups and 3 categories

# Number of categories
n = len(categories)

# Bar width
width = 0.03  # Smaller width to accommodate many bars

# X-axis positions
x = np.arange(n)

# Plotting 20 bars per category
for i in range(n_bars):
    # Adjusting the position for each group
    plt.bar(x + (i - (n_bars // 2)) * width, group_values[i], width=width, label=f'Group {i+1}' if i < 5 else "", align='center')

# Labels and title
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Grouped Bar Chart with 20 Bars per Category')

# Set x-axis labels
plt.xticks(x, categories)

# Add a legend (limit it to the first 5 groups for clarity)
plt.legend(ncol=2)

# Show the plot
plt.show()
```

```python
#Example 30 Student Performance Comparison

import numpy as np
import matplotlib.pyplot as plt

# Data
subjects = ['Math', 'Science', 'English', 'History', 'Computers']
num_students = 30
students = [f'Student {i+1}' for i in range(num_students)]

# Fixed scores for 30 students (each row = 1 student, columns = subjects)
scores = np.array([
    [85, 90, 75, 70, 95], [80, 85, 78, 88, 92], [78, 83, 80, 85, 90],  
    [82, 88, 79, 77, 96], [75, 80, 85, 90, 85], [90, 92, 88, 86, 89],  
    [76, 82, 79, 88, 91], [85, 87, 82, 84, 93], [89, 91, 76, 85, 88],  
    [78, 85, 80, 79, 90], [92, 88, 86, 90, 95], [87, 85, 83, 81, 88],  
    [75, 79, 77, 80, 85], [80, 83, 85, 87, 92], [85, 90, 88, 86, 89],  
    [79, 82, 78, 88, 91], [83, 87, 80, 85, 93], [86, 89, 76, 85, 88],  
    [82, 85, 84, 79, 90], [88, 92, 89, 91, 94], [85, 87, 83, 81, 88],  
    [75, 79, 77, 80, 85], [80, 83, 85, 87, 92], [85, 90, 88, 86, 89],  
    [79, 82, 78, 88, 91], [83, 87, 80, 85, 93], [86, 89, 76, 85, 88],  
    [82, 85, 84, 79, 90], [88, 92, 89, 91, 94], [85, 87, 83, 81, 88]  
])

# Calculate total scores and sort students
total_scores = scores.sum(axis=1)  # Sum of each student's scores
sorted_indices = np.argsort(-total_scores)  # Sort in descending order

# Reorder students and scores based on total score
students = [students[i] for i in sorted_indices]
scores = scores[sorted_indices]

# Bar chart setup
x = np.arange(num_students)  # Use student indices on x-axis
width = 0.15  # Bar width

plt.figure(figsize=(12, 6))

# Plot bars for each subject
for i, subject in enumerate(subjects):
    plt.bar(x + i * width - (len(subjects) / 2) * width, scores[:, i], width, label=subject)

# Labels and title
plt.xlabel('Students (Sorted by Total Score)')
plt.ylabel('Scores')
plt.title('Student Performance Comparison')
plt.xticks(x, students, rotation=90)  # Show student names
plt.legend()  # Show subject-wise legend

plt.show()
```
