```python
# A scatter plot is a type of data visualization used to represent the relationship between two variables. 
# It is a graph where individual data points are plotted on the horizontal (x) and vertical (y) axes. 
# Each point represents a pair of values (x, y), making it easy to identify patterns, correlations, or trends in the data.

# Key Features:
#     X-axis and Y-axis: Represent two different variables.
#     Data points: Each point corresponds to a pair of values, with its position determined by its x-value and y-value.
#     Patterns: Scatter plots are used to observe relationships such as linear or non-linear patterns between the variables.

# When to Use:
#     To visualize the relationship between two continuous variables.
#     To check for correlation (e.g., positive, negative, or no correlation).
#     To identify outliers or unusual data points.
#     To visualize trends or clusters in the data.
```

```python
#Basic scatter plot
```

```python
import matplotlib.pyplot as plt

# Data for the scatter plot
x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 5]

# Create a scatter plot
plt.scatter(x, y)

# Add labels and title
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.title('Simple Scatter Plot')

# Show the plot
plt.show()
```

```python
# Customizing the Scatter Plot:
# You can further customize your scatter plot with different colors, sizes of points, or even adding a grid.
```

```python
plt.scatter(x, y, color='red', s=100)  # Red points with a size of 100

plt.grid(True)  # Add a grid

plt.show()
```

```python
# Example:
# Imagine you want to analyze the relationship between the number of hours studied and the score on a test. You collect data for 5 students:
```

```python
import matplotlib.pyplot as plt

# Data: Hours Studied and Test Scores
x = [1, 2, 3, 4, 5]  # Hours studied
y = [50, 55, 60, 70, 80]  # Test scores

# Create a scatter plot
plt.scatter(x, y, color='pink', s=150)

# Add title and label
plt.title('Scatter Plot: Hours Studied vs Test Score')
plt.xlabel('Hours Studied')
plt.ylabel('Test Score')

# Show the plot
plt.show()
```

```python
# Example 1: Positive Correlation (Linear) : Both variables increase together (e.g., exercise and calories burned).

# Q. Imagine we have data showing the relationship between the number of hours of exercise and the number of calories burned.

# Data:
# Hours of Exercise (X)	Calories Burned (Y)
# 1	                    200
# 2	                    400
# 3	                    600
# 4	                    800
# 5	                    1000
```

```python
import matplotlib.pyplot as plt

Hours_Excercise=[1, 2, 3, 4, 5]
Calories_Burn=[200, 400, 600, 800, 1000]

plt.scatter(Hours_Excercise, Calories_Burn, color='red', s=100)

plt.title('Positive Correlation: Hours of Exercise vs Calories Burned')
plt.xlabel('Hours of Exercise')
plt.ylabel('Calories Burned')

plt.show()
```

```python
# Example 2: Negative Correlation : As one variable increases, the other decreases (e.g., TV watching and books read).

# Q. Now, let’s consider data that shows the relationship between the amount of time spent watching TV and the number of books read. 
# We might expect a negative correlation: as people watch more TV, they read fewer books.

# Data:
# Hours Watching TV (X)	Books Read (Y)
# 1	                    15
# 2	                    12
# 3	                    9
# 4	                    6
# 5	                    3
```

```python
import matplotlib.pyplot as plt

X=[1, 2, 3, 4, 5]
Y=[15, 12, 9, 6, 3]

plt.scatter(X, Y, color='orange')

plt.title('Negative Correlation: Hours Watching TV vs Books Read')
plt.xlabel('Hours Watching TV')
plt.ylabel('Books Read')

plt.show()
```

```python
# Example 3: No Correlation (Random Data) : No visible relationship between the variables (e.g., coffee consumption and phone calls).

# Data:
# Cups of Coffee (X)	Phone Calls Made (Y)
# 1	                8
# 2	                5
# 3	                12
# 4	                9
# 5	                6
```

```python
import matplotlib.pyplot as plt

X=[1, 2, 3, 4, 5]
Y=[8, 5, 12, 9, 6]

plt.scatter(X, Y, c='yellow', marker='d', edgecolor='black')

plt.xlabel('Cups of Coffee')
plt.ylabel('Phone Calls Made')
plt.title('No Correlation: Cups of Coffee vs Phone Calls')

plt.show()
```

```python
# Example 4: Curved Relationship (Non-Linear) : The relationship is not straight but follows a curve (e.g., speed and time taken).

# Q. Let’s explore a case where the relationship between two variables is non-linear. 
# For example, the relationship between the speed of a car and the time it takes to travel a fixed distance.

# Data:
# Speed (X)	Time Taken (Y)
# 10	        60
# 20	        30
# 30	        20
# 40	        15
# 50	        12
```

```python
import matplotlib.pyplot as plt

# Data: Speed vs Time Taken
x = [10, 20, 30, 40, 50]  # Speed (km/h)
y = [60, 30, 20, 15, 12]  # Time taken (minutes)

plt.scatter(x, y, color='purple')

plt.xlabel('Speed (km/h)')
plt.ylabel('Time Taken (minutes)')
plt.title('Non-Linear Relationship: Speed vs Time Taken')

plt.show()
```

```python
# Example 5: Clustered Data (Grouping) : Data points form distinct groups (e.g., age and height across different age ranges).

# Q. In some cases, the data might show clear groupings or clusters of points. 
# Let’s consider data showing the height and weight of people from different age groups.

# Data:
# Age (X)	Height (Y)
# 5	    105
# 6	    110
# 7	    115
# 30	    170
# 32	    175
# 35	    180
# 60	    165
# 62	    160
# 65	    155
```

```python
import matplotlib.pyplot as plt

# Data: Age vs Height
x = [5, 6, 7, 30, 32, 35, 60, 62, 65]  # Age
y = [105, 110, 115, 170, 175, 180, 165, 160, 155]  # Height

plt.scatter(x, y, color='orange')

plt.xlabel('Age')
plt.ylabel('Height')
plt.title('Clustered Data: Age vs Height')

plt.show()
```

```python
# Example 7: Height vs Age (Non-linear Relationship)
# Let’s explore a case showing how height changes with age. 
#Typically, height increases rapidly in childhood and then stabilizes in adulthood, showing a non-linear pattern.

# Data:
# Age (X)	Height (Y)
# 2	        80
# 4	        95
# 6	        110
# 10	    135
# 15	    150
# 20	    160
# 25	    162
```

```python
# Example 8: Advertising Budget vs Sales (Linear Relationship)
# Suppose we have data showing how much a company spends on advertising and the resulting sales revenue. 
#Typically, there is a positive linear relationship between advertising budget and sales.

# Data:
# Advertising Budget (X)	Sales (Y)
# 5000	                    100000
# 10000	                    150000
# 15000	                    200000
# 20000	                    250000
# 25000	                    300000
```

```python
# Example 9: Temperature vs Ice Cream Sales (Non-linear, Seasonal)
# Now, let's explore how ice cream sales increase with temperature. 
#Typically, sales rise with higher temperatures, but at a decreasing rate as the temperature continues to rise, showing a non-linear relationship.

# Data:
# Temperature (X)	Ice Cream Sales (Y)
# 20	            50
# 25	            70
# 30	            90
# 35	            100
# 40	            95
```

```python
import matplotlib.pyplot as plt

x = [20, 25, 30, 35, 40]  # Temperature in Celsius
y = [50, 70, 90, 100, 95]  # Ice cream sales (in units)

plt.scatter(x, y, color='orange')

plt.xlabel('Temperature (°C)')
plt.ylabel('Ice Cream Sales')
plt.title('Non-linear Relationship: Temperature vs Ice Cream Sales')

plt.show()
```

```python
# Example 10: Distance vs Time (Quadratic Relationship)
# Let’s consider a scenario where an object is moving at increasing speed (like a car accelerating), where distance is plotted against time. 
#The relationship between time and distance is quadratic (non-linear).

# Data:
# Time (X)	Distance (Y)
# 1	        2
# 2	        8
# 3	        18
# 4	        32
# 5	        50
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]  # Time in seconds
y = [2, 8, 18, 32, 50]  # Distance in meters

plt.scatter(x, y, color='red')

plt.xlabel('Time (seconds)')
plt.ylabel('Distance (meters)')
plt.title('Quadratic Relationship: Time vs Distance')

plt.show()
```

```python
# Example: study hours and exam scores of boys and girls.

# Student data (study hours and exam scores):

# study_hoursb = [2, 3, 1, 4, 3, 5, 2, 6, 5, 7]
# exam_scoresb = [65, 75, 60, 80, 70, 85, 70, 90, 88, 92]
# study_hoursg = [3, 3, 4, 4, 3, 8, 2, 6, 2, 7]
# exam_scoresg = [75, 70, 60, 80, 70, 90, 70, 90, 88, 100]
```

```python
import matplotlib.pyplot as plt

# Student data (study hours and exam scores)
study_hoursb = [2, 3, 1, 4, 3, 5, 2, 6, 5, 7]
exam_scoresb = [65, 75, 60, 80, 70, 85, 70, 90, 88, 92]
study_hoursg = [3, 3, 4, 4, 3, 8, 2, 6, 2, 7]
exam_scoresg = [75, 70, 60, 80, 70, 90, 70, 90, 88, 100]

# Create a scatter plot
plt.scatter(study_hoursb, exam_scoresb,s=150, c='lightblue', edgecolor='b', marker='^', label='Boys')
plt.scatter(study_hoursg, exam_scoresg,s=150, c='pink', edgecolor='r', marker='d', label='Girls')

# Labeling and Title
plt.xlabel('Study Hours')
plt.ylabel('Exam Scores')
plt.title('Study Hours vs. Exam Scores')

# Show the scatter plot
plt.legend()
plt.grid(True)
plt.show()
```
